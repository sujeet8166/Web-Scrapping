Q1. What is Web Scraping? Why is it Used? Give three areas where Web Scraping is used to get data.

Ans:- Web Scraping is the process of extracting data from websites automatically using software tools, scripts, or programs, It collects information from web pages and converts it into a structured format such as CSV, Excel, or databases.
Why is it Used- Web scraping is used to gather large amounts of data quickly and efficiently from websites, It helps organizations and individuals collect information for analysis, research, and decision-making purposes.
Areas where Web Scraping is Used:-
1. E-commerce and Price Monitoring - Companies collect product prices and compare competitors' pricing strategies.
2. Market Research- Organizations gather customer reviews, ratings, and trends to understand market behavior.
3. News and Content Aggregation- Websites collect news articles and information from multiple sources and display them in one place.

Q2. What are the different methods used for Web Scraping?

Ans:- The different methods used for web scraping are:
1. Manual Copy-Pasting - Data is manually copied from websites and pasted into a file or spreadsheet.
2. HTML Parsing- HTML source code is analyzed to extract required data using programming languages such as Python.
3. DOM Parsing- The Document Object Model (DOM) structure of a webpage is used to locate and retrieve data elements.
4. XPath- XPath is used to navigate through XML and HTML documents to identify specific elements and extract information.
5. Web Scraping Tools and APIs- Tools such as Beautiful Soup, Scrapy, Selenium, and APIs are used to automate data extraction.
6. Selenium-Based Scraping- Selenium is used to scrape dynamic websites that load content using JavaScript.

Q3. What is Beautiful Soup? Why is it used?

Ans:- Beautiful Soup is a Python library used for parsing HTML and XML documents. It creates a parse tree from webpage source code, making it easy to search and extract information.

Why is Beautiful Soup Used?
1. It helps parse HTML and XML documents efficiently.
2. It provides simple methods to navigate web page elements.
3. It makes data extraction easier from web pages.
4. It works well with Python's Requests library.
5. It can handle poorly formatted HTML code.

Example Uses:
Extracting product information, Collecting news headlines, Gathering customer reviews from websites.

Q4. Why is flask used in this Web Scraping project?|

Ans:- Flask is a lightweight Python web framework used to develop web applications.

In a web scraping project, Flask is used for the following purposes:

1. Creating a User Interface- It allows users to enter website URLs or search queries through a web page.
2. Handling User Requests- Flask receives requests from users and sends them to the scraping script.
3. Displaying Scraped Data- It presents the extracted data on web pages in a readable format.
4. Deploying Web Applications- Flask makes it easy to deploy the web scraping project on cloud platforms.
5. Integration with Python- Since Beautiful Soup and other scraping libraries are written in Python, Flask integrates seamlessly with them.
